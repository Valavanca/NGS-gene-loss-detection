```python
from pathlib import Path
import pandas as pd
import json
from src.inference.inference import XGBInference 
```

## Predict and save the results

### Zygosity prediction

```python
model_dir = Path('./models/zygosity_p2_733')
model = XGBInference(model_dir)
model.run_inference(
    './data/val_pids_22052025.csv',
    './output/zyg_pred'
)
```

```python
with open('./output/zyg_pred/predictions.json') as f:
    data = json.load(f)
    zyg_df = pd.DataFrame(data)

zyg_df = zyg_df.rename(columns={'ids': 'AN number', 'preds': 'Zygosity'})
zyg_df = pd.concat([zyg_df.drop(['Zygosity'], axis=1), zyg_df['Zygosity'].apply(pd.Series)], axis=1)
zyg_df.columns = ['AN number', 'Zygosity DRAGEN 0', 'Zygosity DRAGEN 1', 'Zygosity DRAGEN 2']
```
